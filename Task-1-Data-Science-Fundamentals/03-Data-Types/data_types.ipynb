# 3. Data Types

**Definition:** Data types in Data Science describe the nature and structure of data, helping determine how data should be stored, processed, and analyzed.

## 3.1 Numerical Data

**Definition:** Numerical data represents measurable quantities that can be expressed using numbers and used in mathematical calculations.

```python
import pandas as pd

data = pd.DataFrame({
    "Age": [20, 25, 30, 35],
    "Salary": [25000, 35000, 45000, 55000]
})

print(data)
print("\nData Types:")
print(data.dtypes)
```

## 3.2 Categorical Data

**Definition:** Categorical data represents groups or labels that describe different categories or classes.

```python
data = pd.DataFrame({
    "Department": ["IT", "HR", "Sales", "IT"],
    "Status": ["Active", "Active", "Inactive", "Active"]
})

print(data)
print("\nData Types:")
print(data.dtypes)
```

## 3.3 Nominal Data

**Definition:** Nominal data consists of categories that have names or labels but no natural order or ranking.

```python
cities = pd.Series(["Delhi", "Mumbai", "Kolkata", "Delhi"])

print(cities)
print("\nUnique categories:", cities.unique())
```

## 3.4 Ordinal Data

**Definition:** Ordinal data contains categories that have a meaningful order or ranking, although the differences between categories are not necessarily equal.

```python
satisfaction = pd.Categorical(
    ["Low", "Medium", "High", "Medium"],
    categories=["Low", "Medium", "High"],
    ordered=True
)

print(satisfaction)
print("\nOrdered categories:", satisfaction.categories)
```

## 3.5 Boolean Data

**Definition:** Boolean data has two possible values, typically True or False, and is often used to represent binary conditions.

```python
is_approved = pd.Series([True, False, True, True])

print(is_approved)
print("\nData Type:", is_approved.dtype)
```

## 3.6 Datetime Data

**Definition:** Datetime data represents dates and times and is useful for analyzing time-based patterns and trends.

```python
dates = pd.to_datetime([
    "2026-01-10",
    "2026-02-15",
    "2026-03-20"
])

print(dates)
print("\nData Type:", dates.dtype)
```

## 3.7 Structured Data

**Definition:** Structured data follows a predefined format, usually organized into rows and columns with clearly defined fields.

```python
data = pd.DataFrame({
    "Name": ["A", "B", "C"],
    "Age": [21, 22, 23],
    "Score": [85, 90, 88]
})

print(data)
```

## 3.8 Unstructured Data

**Definition:** Unstructured data does not follow a fixed tabular structure and commonly includes text, images, audio, and video.

```python
documents = [
    "Machine learning is a part of AI.",
    "Python is widely used in Data Science.",
    "Data preprocessing improves model quality."
]

for document in documents:
    print(document)
```

## 3.9 Missing Data

**Definition:** Missing data occurs when one or more expected values are unavailable in a dataset.

```python
data = pd.DataFrame({
    "Age": [21, None, 23, 24],
    "Score": [80, 85, None, 90]
})

print(data)
print("\nMissing values:")
print(data.isnull().sum())
```

## 3.10 Data Type Conversion

**Definition:** Data type conversion changes data from one type to another when required for analysis or machine learning.

```python
data = pd.DataFrame({
    "Age": ["21", "22", "23"],
    "Score": ["80", "85", "90"]
})

data["Age"] = data["Age"].astype(int)
data["Score"] = data["Score"].astype(float)

print(data)
print("\nUpdated Data Types:")
print(data.dtypes)
```

## Summary

The examples demonstrate important Data Science data types, including numerical, categorical, nominal, ordinal, boolean, datetime, structured, and unstructured data, along with missing data and type conversion.
