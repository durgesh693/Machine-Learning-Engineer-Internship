# 1. Python Fundamentals

**Definition:** Python fundamentals are the basic concepts and programming principles required to write, understand, and work with Python programs effectively.

## 1.1 Variables

**Definition:** A variable is a name used to store a value in a Python program.

```python
name = "Python"
students = 50
accuracy = 0.92

print(name)
print(students)
print(accuracy)
```

## 1.2 Data Types

**Definition:** A data type specifies the kind of value stored by a variable.

```python
name = "Data Science"       # str
students = 50              # int
accuracy = 0.92            # float
is_completed = True        # bool

print(type(name))
print(type(students))
print(type(accuracy))
print(type(is_completed))
```

## 1.3 Type Conversion

**Definition:** Type conversion is the process of changing a value from one data type to another.

```python
age = "21"
age = int(age)

score = 85
score = float(score)

print(age, type(age))
print(score, type(score))
```

## 1.4 Operators

**Definition:** Operators are symbols or keywords used to perform operations on values and variables.

```python
a = 10
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Modulus:", a % b)
print("Greater than:", a > b)
```

## 1.5 Conditional Statements

**Definition:** Conditional statements execute different blocks of code depending on whether a condition is true or false.

```python
accuracy = 0.92

if accuracy >= 0.90:
    print("Excellent model performance")
elif accuracy >= 0.70:
    print("Good model performance")
else:
    print("Model needs improvement")
```

## 1.6 Loops

**Definition:** Loops repeatedly execute a block of code over a sequence or while a condition remains true.

```python
for model in ["Linear Regression", "Decision Tree", "Random Forest"]:
    print(model)
```

## 1.7 Functions

**Definition:** A function is a reusable block of code designed to perform a specific task.

```python
def calculate_mean(a, b, c):
    return (a + b + c) / 3

result = calculate_mean(70, 80, 90)
print("Mean:", result)
```

## 1.8 Lists

**Definition:** A list is an ordered and mutable collection that can store multiple values.

```python
models = ["Linear Regression", "Decision Tree", "Random Forest"]

print(models)
print("First model:", models[0])

models.append("K-Means")
print("Updated list:", models)
```

## 1.9 Tuples

**Definition:** A tuple is an ordered collection of values that cannot be modified after creation.

```python
model_metrics = (0.92, 0.89, 0.91)

print(model_metrics)
print("First value:", model_metrics[0])
```

## 1.10 Sets

**Definition:** A set is an unordered collection of unique values.

```python
algorithms = {"Regression", "Classification", "Regression", "Clustering"}

print(algorithms)
```

## 1.11 Dictionaries

**Definition:** A dictionary stores data as key-value pairs and allows values to be accessed using keys.

```python
model = {
    "name": "Random Forest",
    "type": "Classification",
    "accuracy": 0.92
}

print(model["name"])
print(model["accuracy"])
```

## 1.12 Exception Handling

**Definition:** Exception handling manages runtime errors so a program can respond to unexpected situations.

```python
try:
    value = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero.")
```

## 1.13 NumPy Basics

**Definition:** NumPy is a Python library used for efficient numerical computing with arrays and mathematical operations.

```python
import numpy as np

marks = np.array([70, 80, 90, 85, 75])

print("Mean:", np.mean(marks))
print("Maximum:", np.max(marks))
print("Minimum:", np.min(marks))
```

## 1.14 Pandas Basics

**Definition:** Pandas is a Python library that provides data structures and tools for data manipulation and analysis.

```python
import pandas as pd

data = {
    "Model": ["Linear Regression", "Decision Tree", "Random Forest"],
    "Accuracy": [0.82, 0.87, 0.92]
}

df = pd.DataFrame(data)

print(df)
print("\nAverage Accuracy:", df["Accuracy"].mean())
```

## Summary

The examples demonstrate core Python fundamentals commonly used in Data Science, including variables, data types, operators, control flow, functions, collections, exception handling, NumPy, and Pandas.
